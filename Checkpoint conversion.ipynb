# Checkpoint conversion from .pth to .safetensors

```python
import torch
from safetensors.torch import save_file
```

```python
#load checkpoints
path = "/home/simon/Documents/Zwischen Wörtern und Pixeln/"

model_swin = path + "SWIN-Base_checkpoint_35epochs_good-results.pth"
model_deit = path + "DEIT-Base_checkpoint_35epochs_good-results.pth"

state_dict_swin = torch.load(model_swin, map_location = "cpu")
state_dict_deit = torch.load(model_deit, map_location = "cpu")
```

```python
#verify
print(type(state_dict_swin))
print(type(state_dict_deit))

#get keys if object is dictionary
if isinstance(state_dict_swin, dict):
    print(state_dict_swin.keys())

if isinstance(state_dict_deit, dict):
    print(state_dict_deit.keys())
```

```python
#save as .safetensors
save_file(state_dict_swin, path + "SWIN-Base_checkpoint_final.safetensors")
save_file(state_dict_deit, path + "DEIT-Base_checkpoint_final.safetensors")
```

```python
from safetensors.torch import load_file


tensors_swin = load_file(path + "SWIN-Base_checkpoint_final.safetensors")
tensors_deit = load_file(path + "DEIT-Base_checkpoint_final.safetensors")

#verify that tensors are identical
for tensor in tensors_swin:
    assert torch.equal(tensors_swin[tensor],
                       state_dict_swin[tensor]
                      )

for tensor in tensors_deit:
    assert torch.equal(tensors_deit[tensor],
                        state_dict_deit[tensor]
                    )

print("Validation completed")
```
